```python
from ultralytics import RTDETR

# 1. Load your best fine-tuned model checkpoint
model_path = "/kaggle/working/road_damage_project/rtdetr_run1-2/weights/best.pt"
model = RTDETR(model_path)

# 2. Run validation specifically on the 'test' split defined in data.yaml
print("Running evaluation on the test split...")
metrics = model.val(
    data="/kaggle/working/data.yaml",
    split="test",          # Forces evaluation on images/test
    batch=8,
    imgsz=640,
    device=0,
    plots=True             # Generates confusion matrices and sample predictions
)

# 3. Print Overall Metrics
print("\n" + "=" * 45)
print("             OVERALL TEST METRICS            ")
print("=" * 45)
print(f"Mean Precision (P) : {metrics.box.mp:.4f}")
print(f"Mean Recall (R)    : {metrics.box.mr:.4f}")
print(f"mAP@0.50           : {metrics.box.map50:.4f}")
print(f"mAP@0.50:0.95      : {metrics.box.map:.4f}")
print("=" * 45)

# 4. Print Per-Class Breakdown
print("\n" + "=" * 45)
print("              PER-CLASS METRICS              ")
print("=" * 45)
for cls_id, cls_name in model.names.items():
    p, r, ap50, ap = metrics.box.class_result(cls_id)
    print(f"Class [{cls_name}]:")
    print(f"  Precision: {p:.4f} | Recall: {r:.4f} | mAP50: {ap50:.4f} | mAP50-95: {ap:.4f}")
print("=" * 45)
```

```python
import cv2
import matplotlib.pyplot as plt
from ultralytics import RTDETR

# 1. Load the model
model_path = "/kaggle/working/road_damage_project/rtdetr_run1-2/weights/best.pt"
model = RTDETR(model_path)

# 2. Path to an image from your test set
# (Make sure this filename matches an actual file in your images/test directory)
image_path = "/kaggle/input/datasets/dhoesh123/image1/images (1).jpg"

# 3. Run inference
results = model.predict(source=image_path, conf=0.25, imgsz=640, device=0)
result = results[0]

# 4. Print detections cleanly
print("\n" + "=" * 50)
print(f"Detections for: {image_path}")
print("=" * 50)

boxes = result.boxes
if len(boxes) == 0:
    print("No damages detected above the confidence threshold.")
else:
    for i, box in enumerate(boxes):
        cls_id = int(box.cls[0].item())
        cls_name = model.names[cls_id]
        confidence = float(box.conf[0].item())
        coords = [round(x, 2) for x in box.xyxy[0].tolist()]

        print(f"Detection #{i + 1}:")
        print(f"  Class      : {cls_name} (ID: {cls_id})")
        print(f"  Confidence : {confidence:.4f} ({confidence * 100:.1f}%)")
        print(f"  BBox (xyxy): {coords}")
        print("-" * 30)

# 5. Display image with bounding boxes
annotated_img = result.plot()
annotated_img_rgb = cv2.cvtColor(annotated_img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(8, 8))
plt.imshow(annotated_img_rgb)
plt.axis("off")
plt.title(f"Detections: {len(boxes)} found")
plt.show()
```
